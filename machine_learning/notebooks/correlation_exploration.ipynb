**Topics by Campus / Level / Genre**

```python
### Distribution of topics by Campus
topic_by_campus = (
    analysis_topics.groupby(["Campus", "topic"])
    .size()
    .groupby(level=0)
    .apply(lambda s: (s / s.sum()).sort_values(ascending=False).head(5))
    .rename("Top 5 topics on this campus")
)
```

```python
### Distribution of topics by level
topic_by_level = (
    analysis_topics.groupby(["AcademicLevel", "topic"])
    .size()
    .groupby(level=0)
    .apply(lambda s: (s / s.sum()).sort_values(ascending=False).head(5))
    .rename("Top 5 topics on this level")
)
```

```python
### Distribution of topics by genre
topic_by_genre = (
    analysis_topics.groupby(["Genre", "topic"])
    .size()
    .groupby(level=0)
    .apply(lambda s: (s / s.sum()).sort_values(ascending=False).head(5))
    .rename("Top 5 topics on this genre")
)
```
